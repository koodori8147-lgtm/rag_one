```python
from langchain_community.document_loaders import CSVLoader
loader = CSVLoader(
    file_path="./data/titanic.csv",
    csv_args={
        "delimiter": ",",
        "quotechar": '"',
        "fieldnames": [
            "Passenger ID",
            "Survival (1: Survived, 0:Died)",
            "Passenger Class",
            "Name",
            "Sex",
            "Age",
            "Number of Siblings/Spouses Aboard",
            "Number of Parents/Children Aboard",
            "Ticket Number",
            "Fare",
            "Cabin",
            "Port of Embarkation",
        ],
    },
)

docs = loader.load()
print(docs[15].page_content)
```

```python
row = docs[15].page_content.split("\n")
row_str = "<row>"
for element in row:
    splitted_element = element.split(":")
    value = splitted_element[-1]
    col = ":".join(splitted_element[:-1])
    row_str+=f"<{col}>{value.strip()}</{col}>"
row_str += "</row>"
print(row_str)
```

```python
for doc in docs:
    row = doc.page_content.split("\n")
    row_str="<row>"
    for element in row:
        splitted_element = element.split(":")
        value = splitted_element[-1]
        col = ":".join(splitted_element[:-1])
        row_str += f"<{col}>{value.strip()}</{col}>"
    row_str += "</row>"
    print(row_str)

loader = CSVLoader(
    file_path="./data/titanic.csv", source_column="Name"
)

docs = loader.load()
print(docs[15])
```

```python
import bs4
from langchain_community.document_loaders import WebBaseLoader

loader = WebBaseLoader(
    web_paths=("https://www.insight.co.kr/news/574195",),
)

```

```python
loader = WebBaseLoader(
    web_paths=("https://www.insight.co.kr/news/574195",),

    bs_kwargs=dict(
        parse_only=bs4.SoupStrainer(
            class_=[
                "main-news-header-title",
                "main-news-article-body"
            ]
        )
    ),

    header_template={
        "User-Agent": "Mozilla/5.0"
    },
)

docs = loader.load()
print(f"문서의 수: {len(docs)}")
docs
```
